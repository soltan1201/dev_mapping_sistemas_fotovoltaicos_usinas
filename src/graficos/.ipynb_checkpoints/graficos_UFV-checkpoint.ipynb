# Gráficos de usinas fotovoltaicas por tipo de região
Leitura de `UFV_consolidado.csv` e geração de gráficos por `tipo_region` (bioma, estado e país), no estilo dos gráficos de referência em `figuras/`.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('UFV_consolidado.csv')
df.head()
```

```python
import numpy as np
import matplotlib.colors as mcolors
from matplotlib.ticker import FuncFormatter
import textwrap

# Paleta sequencial usada em todos os gráficos (clara = valor menor/ano mais
# antigo, escura = valor maior/ano mais recente) — mesmo tom de laranja/marrom
# dos gráficos de referência em figuras/.
CMAP_LARANJA = mcolors.LinearSegmentedColormap.from_list(
    'ufv_laranja', ['#fbe0c4', '#f2a860', '#d9702f', '#b23a2f', '#7a2618']
)


def rampa_cores(n: int, inverso: bool = False):
    """`n` cores igualmente espaçadas na paleta sequencial (clara → escura)."""
    posicoes = np.linspace(0.05, 0.95, n)
    if inverso:
        posicoes = posicoes[::-1]
    return [CMAP_LARANJA(p) for p in posicoes]


def formata_ptbr(valor, casas: int = 0) -> str:
    """Formata número no padrão brasileiro (ponto como separador de milhar)."""
    s = f'{valor:,.{casas}f}'
    return s.replace(',', '_').replace('.', ',').replace('_', '.')


def quebra_nome(nome: str, largura: int = 12) -> str:
    """Quebra nomes de região longos em 2 linhas para caber no eixo X."""
    return '\n'.join(textwrap.wrap(nome, width=largura, break_long_words=False))


FORMATADOR_PTBR = FuncFormatter(lambda v, _: formata_ptbr(v))

print('Utilitários de estilo definidos.')
```

## Biomas
Gráfico de barras verticais com o **ano no eixo X** e uma barra por bioma, seguido de um gráfico de pizza com a participação de cada bioma no ano mais recente — estilo do gráfico de referência em `figuras/pocentagem_UFV_biomas.png`.

```python
from matplotlib.scale import FuncScale

df_bioma = df[df['tipo_region'] == 'bioma']
pivot_bioma = df_bioma.pivot_table(index='year', columns='nome_region', values='area_ha', aggfunc='sum')

# escala não-linear por trechos, para não perder a variação dos biomas menores
# frente aos valores muito maiores da Caatinga
y_breaks = [0, 500, 2500, 15000, 30000]
y_positions = [0, 1, 2, 3, 4]

def forward(y):
    return np.interp(y, y_breaks, y_positions)

def inverse(y):
    return np.interp(y, y_positions, y_breaks)

ax = pivot_bioma.plot(kind='bar', figsize=(12, 6), width=0.9)
ax.set_yscale(FuncScale(ax.yaxis, (forward, inverse)))
ax.set_yticks(y_breaks)
ax.yaxis.set_major_formatter(FORMATADOR_PTBR)
ax.set_title('Área de usinas fotovoltaicas por bioma e ano')
ax.set_xlabel('Ano')
ax.set_ylabel('Área (ha)')
ax.legend(title='Bioma')

# rótulo com o % da área do ano que é da Caatinga, só na barra da Caatinga
pct_caatinga = pivot_bioma['Caatinga'] / pivot_bioma.sum(axis=1) * 100
container_caatinga = ax.containers[list(pivot_bioma.columns).index('Caatinga')]
ax.bar_label(container_caatinga, labels=[f'{p:.0f}%' for p in pct_caatinga], padding=3)

plt.tight_layout()
plt.show()
```

### Participação por bioma (ano mais recente)
Percentual da área de usinas fotovoltaicas por bioma no último ano disponível.

```python
ultimo_ano_bioma = pivot_bioma.index.max()
serie_biomas = pivot_bioma.loc[ultimo_ano_bioma].sort_values(ascending=False)
total_biomas = serie_biomas.sum()
cores_pizza = rampa_cores(len(serie_biomas), inverso=True)

fig, ax = plt.subplots(figsize=(7, 6.5))
wedges, _ = ax.pie(
    serie_biomas.values,
    colors=cores_pizza,
    startangle=90,
    counterclock=False,
    wedgeprops=dict(edgecolor='white', linewidth=1.5),
)

LIMIAR_INTERNO = 3.0  # % mínimo para o rótulo caber dentro da fatia
raio_fora = 1.25
for wedge, valor in zip(wedges, serie_biomas.values):
    pct = valor / total_biomas * 100
    ang = np.radians((wedge.theta1 + wedge.theta2) / 2)
    pct_str = f'{pct:.0f}%' if pct >= 1 else f'{pct:.2f}'.replace('.', ',') + '%'
    texto = f'{pct_str}\n{formata_ptbr(valor)} ha'

    if pct >= LIMIAR_INTERNO:
        x, y = 0.65 * np.cos(ang), 0.65 * np.sin(ang)
        ax.annotate(texto, xy=(x, y), ha='center', va='center',
                    fontsize=11, fontweight='bold', color='#2a1408')
    else:
        x_borda, y_borda = np.cos(ang), np.sin(ang)
        x_fora, y_fora = raio_fora * np.cos(ang), raio_fora * np.sin(ang)
        ax.annotate(
            texto, xy=(x_borda, y_borda), xytext=(x_fora, y_fora),
            ha='center', va='center', fontsize=10, fontweight='bold',
            arrowprops=dict(arrowstyle='-', color='#444', lw=1),
        )

# margem extra no topo para o rótulo "de fora" (fatias bem pequenas, perto de
# 90°) não colidir com o título
ax.set_aspect('equal')
ax.set_xlim(-1.45, 1.45)
ax.set_ylim(-1.1, 1.6)
ax.set_title(f'Participação por bioma — {ultimo_ano_bioma}', pad=30)
ax.legend(serie_biomas.index, loc='upper center', bbox_to_anchor=(0.5, -0.02),
          ncol=len(serie_biomas), frameon=False)
plt.tight_layout()
plt.show()
```

## Estados
Gráfico de barras verticais — um grupo por estado (ordenado do maior para o menor pela área do ano mais recente) e uma barra por ano dentro do grupo, com rótulo no ano mais recente — estilo do gráfico de referência em `figuras/grafico_historico_usinas.png`.

```python
df_estado = df[df['tipo_region'] == 'estado']
pivot_estado = df_estado.pivot_table(index='year', columns='nome_region', values='area_ha', aggfunc='sum')

# ordena os estados do maior para o menor tendo como referência o ano mais recente
ultimo_ano_estado = pivot_estado.index.max()
ordem_desc = pivot_estado.loc[ultimo_ano_estado].sort_values(ascending=False).index
pivot_estado = pivot_estado[ordem_desc]

anos = pivot_estado.index.tolist()
estados = pivot_estado.columns.tolist()
cores_anos = rampa_cores(len(anos))

x = np.arange(len(estados))
largura_barra = 0.8 / len(anos)

fig, ax = plt.subplots(figsize=(15, 7))
for i, ano in enumerate(anos):
    valores = pivot_estado.loc[ano].values
    posicoes = x + (i - (len(anos) - 1) / 2) * largura_barra
    container = ax.bar(posicoes, valores, width=largura_barra, color=cores_anos[i], label=str(ano))
    if ano == ultimo_ano_estado:
        ax.bar_label(container, labels=[formata_ptbr(v) for v in valores],
                     padding=3, fontsize=9, fontweight='bold')

ax.set_xticks(x)
ax.set_xticklabels([quebra_nome(e) for e in estados])
ax.yaxis.set_major_formatter(FORMATADOR_PTBR)
ax.set_title('Área de usinas fotovoltaicas por estado e ano')
ax.set_ylabel('Área (ha)')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=len(anos), frameon=False)
ax.spines[['top', 'right']].set_visible(False)
ax.grid(axis='y', color='#e5e5e5', linewidth=0.8, zorder=0)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

## País (Brasil)
Como há apenas uma região do tipo `pais` (Brasil), o gráfico mostra a área por ano — estilo do gráfico de referência em `figuras/historico_UFV_Br.png`.

```python
from matplotlib.collections import LineCollection

df_pais = df[df['tipo_region'] == 'pais']
pivot_pais = df_pais.pivot_table(index='year', columns='nome_region', values='area_ha', aggfunc='sum')
serie_brasil = pivot_pais['Brasil']

anos = serie_brasil.index.to_numpy()
valores = serie_brasil.to_numpy()

# segmentos de reta entre pontos consecutivos, coloridos em gradiente
pontos = np.array([anos, valores]).T.reshape(-1, 1, 2)
segmentos = np.concatenate([pontos[:-1], pontos[1:]], axis=1)
cores_linha = [CMAP_LARANJA(0.1 + 0.85 * p) for p in np.linspace(0, 1, len(anos) - 1)]
cores_pontos = [CMAP_LARANJA(0.1 + 0.85 * p) for p in np.linspace(0, 1, len(anos))]

fig, ax = plt.subplots(figsize=(9, 5.5))
lc = LineCollection(segmentos, colors=cores_linha, linewidth=3, zorder=2)
ax.add_collection(lc)
ax.scatter(anos, valores, color=cores_pontos, s=55, zorder=3, edgecolor='white', linewidth=1.2)

for ano, valor in zip(anos, valores):
    ax.annotate(formata_ptbr(valor), xy=(ano, valor), xytext=(0, 10),
                textcoords='offset points', ha='center', fontsize=9, color='#555')

ax.set_xlim(anos.min() - 0.5, anos.max() + 0.5)
ax.set_ylim(0, valores.max() * 1.18)
ax.set_xticks(anos)
ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: formata_ptbr(v / 1000)))
ax.set_title('Área de usinas fotovoltaicas no Brasil por ano')
ax.set_ylabel('Área (ha)\nMilhares')
ax.spines[['top', 'right']].set_visible(False)
ax.grid(axis='y', color='#e5e5e5', linewidth=0.8, zorder=0)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
