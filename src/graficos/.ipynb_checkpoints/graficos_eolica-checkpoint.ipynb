# Gráficos de energia eólica por tipo de região
Leitura de `eolica_consolidado.csv` e geração de gráficos de barras por `tipo_region` (bioma, estado e país).

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('eolica_consolidado.csv')
df.head()
```

## Biomas
Gráfico de barras verticais com o **ano no eixo X** e 4 barras por ano (uma para cada bioma).

```python
import numpy as np
from matplotlib.scale import FuncScale

df_bioma = df[df['tipo_region'] == 'bioma']
pivot_bioma = df_bioma.pivot_table(index='year', columns='nome_region', values='area_ha', aggfunc='sum')

# escala não-linear por trechos, para não perder a variação dos biomas menores
# frente aos valores muito maiores da Caatinga
y_breaks = [0, 500, 2500, 15000, 30000]
y_positions = [0, 1, 2, 3, 4]

def forward(y):
    return np.interp(y, y_breaks, y_positions)

def inverse(y):
    return np.interp(y, y_positions, y_breaks)

ax = pivot_bioma.plot(kind='bar', figsize=(12, 6), width=0.9)
ax.set_yscale(FuncScale(ax.yaxis, (forward, inverse)))
ax.set_yticks(y_breaks)
ax.set_title('Área de energia eólica por bioma e ano')
ax.set_xlabel('Ano')
ax.set_ylabel('Área (ha)')
ax.legend(title='Bioma')

# rótulo com o % da área do ano que é da Caatinga, só na barra da Caatinga
pct_caatinga = pivot_bioma['caatinga'] / pivot_bioma.sum(axis=1) * 100
container_caatinga = ax.containers[list(pivot_bioma.columns).index('caatinga')]
ax.bar_label(container_caatinga, labels=[f'{p:.0f}%' for p in pct_caatinga], padding=3)

plt.tight_layout()
plt.show()
```

## Estados
Gráfico de barras horizontal (o eixo X vira o eixo Y) para que as barras de todos os estados caibam confortavelmente.

```python
import numpy as np
from matplotlib.scale import FuncScale

df_estado = df[df['tipo_region'] == 'estado']
pivot_estado = df_estado.pivot_table(index='year', columns='nome_region', values='area_ha', aggfunc='sum')

# ordena os estados do menor para o maior tendo como referência a área de 2025
ordem_2025 = pivot_estado.loc[2025].sort_values().index
pivot_estado = pivot_estado[ordem_2025]

# escala não-linear por trechos, para não perder a variação dos estados menores
# frente aos valores muito maiores da Bahia
x_breaks = [0, 100, 250, 1500, 10000]
x_positions = [0, 1, 2, 3, 4]

def forward(x):
    return np.interp(x, x_breaks, x_positions)

def inverse(x):
    return np.interp(x, x_positions, x_breaks)

ax = pivot_estado.plot(kind='barh', figsize=(12, 16), colormap='tab20', width=0.85)
ax.set_xscale(FuncScale(ax.xaxis, (forward, inverse)))
ax.set_xticks(x_breaks)
ax.set_title('Área de energia eólica por estado e ano')
ax.set_xlabel('Área (ha)')
ax.set_ylabel('Ano')
ax.legend(title='Estado', bbox_to_anchor=(1.02, 1), loc='upper left')

# rótulo com o % da área do ano, apenas nas barras com mais de 1000 ha
total_por_ano = pivot_estado.sum(axis=1)
for estado, container in zip(pivot_estado.columns, ax.containers):
    valores = pivot_estado[estado]
    pct = valores / total_por_ano * 100
    labels = [f'{p:.0f}%' if v > 1000 else '' for v, p in zip(valores, pct)]
    ax.bar_label(container, labels=labels, padding=3, fontsize=8)

plt.tight_layout()
plt.show()
```

## País (Brasil)
Como há apenas uma região do tipo `pais` (Brasil), o gráfico mostra a área por ano.

```python
import numpy as np
import plotly.graph_objects as go

df_pais = df[df['tipo_region'] == 'pais']
pivot_pais = df_pais.pivot_table(index='year', columns='nome_region', values='area_ha', aggfunc='sum')
serie_brasil = pivot_pais['Brasil']

anos = serie_brasil.index.to_numpy()
valores = serie_brasil.to_numpy()

# linha de tendência (regressão linear)
coef = np.polyfit(anos, valores, 1)
tendencia = np.polyval(coef, anos)

crescimento_pct = (valores[-1] / valores[0] - 1) * 100

fig = go.Figure()

fig.add_bar(
    x=anos,
    y=valores,
    name='Área (ha)',
    marker_color='#2c5f7c',
    text=[f'{v:,.0f}'.replace(',', '.') for v in valores],
    textposition='outside',
    textfont=dict(size=11),
)

fig.add_scatter(
    x=anos,
    y=tendencia,
    mode='lines',
    name='Tendência linear',
    line=dict(color='#b23a2f', dash='dash', width=2),
)

fig.add_annotation(
    x=anos[-1], y=valores[-1],
    text=f'+{crescimento_pct:.0f}% desde 2016',
    showarrow=True,
    arrowhead=2,
    ax=-70, ay=-45,
    font=dict(size=13, family='Georgia, serif', color='#b23a2f'),
    bgcolor='rgba(255,255,255,0.85)',
    bordercolor='#b23a2f',
    borderwidth=1,
)

fig.update_layout(
    title=dict(
        text='Área de energia eólica no Brasil por ano',
        font=dict(family='Georgia, serif', size=18),
        x=0.5,
    ),
    xaxis_title='Ano',
    yaxis_title='Área (ha)',
    font=dict(family='Georgia, serif', size=12, color='#333'),
    plot_bgcolor='white',
    paper_bgcolor='white',
    yaxis=dict(showgrid=True, gridcolor='#e0e0e0', gridwidth=1, zeroline=False, range=[0, max(valores) * 1.15]),
    xaxis=dict(showgrid=False, tickmode='linear', dtick=1),
    legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1),
    width=900,
    height=550,
    margin=dict(t=90, l=70, r=40, b=60),
)

fig.show()
```
